```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

temperatures = np.arange(0.5, 2.1, 0.1)

final_energy = []
final_energy_err = []
final_energy_G = []
final_energy_G_err = []

for i, t in enumerate(temperatures):
    energy, error_e = np.loadtxt(f"../OUTPUT/total_energy_Metro_h0_{t:.1f}.dat", usecols = (2, 3), unpack = True)
    energy_G, error_e_G = np.loadtxt(f"../OUTPUT/total_energy_Gibbs_h0_{t:.1f}.dat", usecols = (2, 3), unpack = True)
    final_energy.append(energy[-1])
    final_energy_err.append(error_e[-1])
    final_energy_G.append(energy_G[-1])
    final_energy_G_err.append(error_e_G[-1])

points=100
T = np.linspace(0.5,2.0,num=points)
beta = 1/T
J = 1.0
Ns = 50
th = np.tanh(J/T)
thN= th**Ns
ch = 1/th
e = -J*( th + ch*thN )/( 1 + thN )
plt.plot(T, e, label = "Theory")

plt.errorbar(temperatures, final_energy, yerr=final_energy_err, fmt='o', capsize=4, elinewidth=1.5, markersize=4, ecolor='red', label = "Metropolis")
plt.errorbar(temperatures, final_energy_G, yerr=final_energy_G_err, fmt='o', capsize=4, elinewidth=1.5, markersize=4, ecolor='red', label = "Gibbs")

plt.grid(True)
plt.title("Total energy vs Temperature")
plt.xlabel("Temperature (K)")
plt.ylabel("Total energy")
plt.legend()
plt.show()
```

```python
final_cv = []
final_cv_err = []
final_cv_G = []
final_cv_G_err = []


for i, t in enumerate(temperatures):
    cv, error_cv = np.loadtxt(f"../OUTPUT/specific_heat_Metro_h0_{t:.1f}.dat", usecols = (2, 3), unpack = True)
    cv_G, error_cv_G = np.loadtxt(f"../OUTPUT/specific_heat_Gibbs_h0_{t:.1f}.dat", usecols = (2, 3), unpack = True)
    final_cv.append(cv[-1])
    final_cv_err.append(error_cv[-1])
    final_cv_G.append(cv_G[-1])
    final_cv_G_err.append(error_cv_G[-1])

heat=((beta*J)**2)*(((1+thN+(Ns-1)*(th**2)+(Ns-1)*(ch**2)*thN)/(1+thN))-Ns*((th+ch*thN)/(1+thN))**2)
plt.plot(T, heat, label = "Theory")

plt.errorbar(temperatures, final_cv, yerr= final_cv_err, fmt='o', capsize=4, elinewidth=1.5, markersize=4, ecolor='red', label = "Metropolis")
plt.errorbar(temperatures, final_cv_G, yerr= final_cv_G_err, fmt='o', capsize=4, elinewidth=1.5, markersize=4, ecolor='red', label = "Gibbs")
plt.grid(True)
plt.title("Specific heat vs Temperature")
plt.xlabel("Temperature (K)")
plt.ylabel("Specific heat")
plt.legend()
plt.show()
```

```python
final_chi = []
final_chi_err = []
final_chi_G = []
final_chi_G_err = []

for i, t in enumerate(temperatures):
    chi, error_chi = np.loadtxt(f"../OUTPUT/susceptibility_Metro_h0_{t:.1f}.dat", usecols = (2, 3), unpack = True)
    final_chi.append(chi[-1])
    final_chi_err.append(error_chi[-1])
    chi, error_chi = np.loadtxt(f"../OUTPUT/susceptibility_Gibbs_h0_{t:.1f}.dat", usecols = (2, 3), unpack = True)
    final_chi_G.append(chi[-1])
    final_chi_G_err.append(error_chi[-1])

X = beta*np.exp(2*beta*J)*(1-thN)/(1+thN)
plt.plot(T, X, label = "Theory")

plt.errorbar(temperatures, final_chi, yerr= final_chi_err, fmt='o', capsize=4, elinewidth=1.5, markersize=4, ecolor='red', label = "Metropolis")
plt.errorbar(temperatures, final_chi_G, yerr= final_chi_G_err, fmt='o', capsize=4, elinewidth=1.5, markersize=4, ecolor='red', label = "Gibbs")
plt.grid(True)
plt.title("Susceptibility vs Temperature")
plt.xlabel("Temperature (K)")
plt.ylabel("Susceptibility")
plt.legend()
plt.plot()
```
