```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
steps = 250

block, energy, error = np.loadtxt("../OUTPUT/Exercise_04.3/potential_energy(1).dat", usecols = (0, 1, 3), unpack = True)
blockt, t, errort = np.loadtxt("../OUTPUT/Exercise_04.3/temperature(1).dat", usecols = (0, 1, 3), unpack = True)
print(t[19])
plt.plot(block*steps, energy, label = "Potential energy")
plt.plot(block*steps, t, label = "Temperature")
plt.grid(True)
plt.title("Evolution of potential energy and temperature")
plt.xlabel("Blocks")
plt.ylabel("Potential energy and temperature")
plt.plot()
```
