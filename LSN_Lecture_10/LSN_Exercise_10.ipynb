La parallelizzazione è stata compiuta su $9$ core, sul mio portatile.
```python
import numpy as np
import matplotlib.pyplot as plt

plt.rc('font', size=14)
```

```python
file_prefix = "Exercise_10.1/circle_metrics_core_"
num_cores = 9  

plt.figure(figsize=(12, 8))

for i in range(num_cores):
    filename = f"{file_prefix}{i}.dat"
    
    data = np.loadtxt(filename)
        
    generations = data[:, 0]
    best_length = data[:, 1]
        
    plt.plot(generations, best_length, label=f'Core {i}', alpha=0.8)

plt.title('Evoluzione del miglior percorso per ogni core')
plt.xlabel('Generazione')
plt.ylabel('Lunghezza del percorso')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

# <span style="color:red"> Migrazioni di tipo 1</span>

Le migrazioni di tipo $1$ avvengono con  il migliore di ogni continente che migra verso un altro continente, **sostituendo un individuo casuale che non sia il migliore del continente destinatario**.

Il TSP è stato affrontato con computazione parallela a 9 core. Ogni 100 generazioni, il migliore di ogni continente migra verso un altro continente, **sostituendo un individuo casuale che non sia il migliore del continente destinatario**.

Dopo 10000 generazioni, viene costituita una popolazione finale, con i migliori di ogni continente (100 individui finali per 9 core, quindi i migliori 11 individui di ogni core e i migliori 12 del core 0). Questa popolazione dell'"eden" subisce evoluzione per ulteriori 200 generazioni. 

Sono riportati:
- L'evoluzione della lunghezza del percorso migliore di ogni rank in funzione delle generazioni;
- L'evoluzione della lunghezza del percorso migliore e della medie della migliore metà per le ultime 200 generazioni dopo la selezione finale;
- Il percorso migliore alla 200esima e ultima generazione.

```python
file_prefix = "parallel_migrations_1/migrations_metrics_core_"
num_cores = 9  

plt.figure(figsize=(12, 8))

for i in range(num_cores):
    filename = f"{file_prefix}{i}.dat"
    
    data = np.loadtxt(filename)
        
    generations = data[:, 0]
    best_length = data[:, 1]
        
    plt.plot(generations, best_length, label=f'Core {i}', alpha=0.8)

plt.title('Evoluzione del miglior percorso per ogni core')
plt.xlabel('Generazione')
plt.ylabel('Lunghezza del percorso')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
file_prefix = "parallel_migrations_1/migrations_metrics_core_"
fig, axes = plt.subplots(2, 1, figsize = (12, 14))

for i in range(num_cores):
    filename = f"{file_prefix}{i}.dat"
    
    data = np.loadtxt(filename)
        
    generations = data[:, 0]
    best_length = data[:, 1]
        
    axes[0].plot(generations[0:1000], best_length[0:1000], label=f'Core {i}', alpha=0.8)
    axes[1].plot(generations[9500:10000], best_length[9500:10000], label=f'Core {i}', alpha=0.8)


axes[0].set_title(r"Prime 1000 generazioni")
axes[0].set_xlabel("Generazioni")
axes[0].set_ylabel("Lunghezza Percorso")
axes[0].legend()
axes[0].grid(True, linestyle=':', alpha=0.7)


axes[1].set_title(r"Ultime 500 generazioni")
axes[1].set_xlabel("Generazioni")
axes[1].set_ylabel("Lunghezza Percorso")
axes[1].legend()
axes[1].grid(True, linestyle=':', alpha=0.7)

plt.suptitle('Evoluzione del miglior percorso per ogni core')

plt.show()
```

Possiamo osservare come per una mappa strutturalmente più complessa, l'algoritmo sia più lento a convergere al minimo. Allo stesso tenmpo, però, con questo sistema di migrazioni il codice ha univocamente raggiunto il minimo con tutti i core prima che si arrivi all'evoluzione finale

```python
generations, best_length, mean_length = np.loadtxt('parallel_migrations_1/parallel_metrics_eden.dat', unpack=True)

plt.figure(figsize=(10, 6))
print(f"Lunghezza del miglior percorso all'inizio dell'evoluzione finale", best_length[0])
print(f"Lunghezza del miglior percorso alla fine dell'evoluzione finale", best_length[-1])

plt.plot(generations, best_length, label=r'Miglior Percorso ($L_{best}$)', color='dodgerblue', linewidth=2)

plt.plot(generations, mean_length, label=r'Media miglior metà ($\langle L \rangle$)', color='crimson', linestyle='--', linewidth=2)

plt.title(r"Funzioni costo durante l'evoluzione finale")
plt.xlabel("Generazioni")
plt.ylabel("Lunghezza Percorso")
plt.legend()
plt.grid(True, linestyle=':', alpha=0.7)

plt.show()
```

L'evoluzione del continente degli "eletti" non sembra utilisima: la lunghezza del percorso migliore non diminuisce, sembra che il sistema abbia già raggiunto il minimo e per effetto delle mutazioni oscilli a ridosso di questo valore nella media della migliore metà. Non escludo, però, che allungare la durata di quest'ultimo periodo di evoluzione possa portare benefici.

Un'altra idea potrebbe essere quella di aumentare la frequenza delle mutazioni: all'inizio dell'ultima selezione gli individui sono molto simili tra loro, come testimonia il fatto che la media della migliore metà sia sovrapposta alla lunghezza del percorso migliore. Per questo motivo, credo che rendere le mutazioni più frequenti non possa arrecare troppo danno all'algoritmo, che conserverebbe sempre svariate copie del percorso migliore, ma guadagnerebbe in diversità e probabilità di uscire da eventuali minimi locali.

```python
x, y = np.loadtxt('parallel_migrations_1/parallel_best_path.dat', unpack=True)
print(f"Lunghezza del miglior percorso", best_length[-1])


plt.figure(figsize=(8, 8))

plt.plot(x, y, marker='o', markersize=6, linestyle='-', zorder = 1, color='teal')
plt.scatter(x, y, color='orange', s=50, zorder = 3, label='Percorso')

plt.plot(x[0], y[0], marker='*', markersize=15, color='gold', markeredgecolor='black', zorder = 10, label='Città 0 (Partenza/Arrivo)')

# Aggiunge i numeri delle città vicino ai punti
#for i in range(len(x)-1):
#    plt.text(x[i], y[i], f" {i}", fontsize=15, color='darkred')

plt.title("Miglior percorso")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True, linestyle=':', alpha=0.7)

#plt.axis('equal') 

plt.show()
```

Il percorso trovato sembra ragionevole: a occhio non vedo particolari errori, forse non è ottimale la gestione della alpi e della Bassa padana. Caratteristica è la scelta di risalire la dorsale Nord-Sud zig-zagando tra il Tirreno e l'Adriatico. Mi sembra ragionevole, ma osserveremo che l'algoritmo successivo ha fatto altre scelte, ottenendo un risultato leggermente migliore.
