```python
import numpy as np
import matplotlib.pyplot as plt

# ====================================================================
# PARAMETRI DELLA SIMULAZIONE
# ====================================================================
N_part = 108   # Numero di particelle
n_bins = 30    # Numero di bin dell'istogramma
mass = 1.0     

# ====================================================================
# 1. TROVARE LA TEMPERATURA EFFETTIVA (T_eff)
# ====================================================================

temp_data = np.loadtxt("../OUTPUT/temperature4.1.dat", usecols = 1, unpack = True) 


# ====================================================================
# 2. CARICARE E RIMODELLARE I DATI DELL'ISTOGRAMMA
# ====================================================================
v, hist_prog, err_prog = np.loadtxt("../OUTPUT/pofv4.1.dat", usecols = (0, 2, 3), unpack = True)

n_blocks = 20
print(f"Letti {n_blocks} blocchi dal file pofv4.1.dat.")
bin_size = v[1] - v[0]

v_blocks = v.reshape(n_blocks, n_bins)
hist_prog_blocks = hist_prog.reshape(n_blocks, n_bins)
err_prog_blocks = err_prog.reshape(n_blocks, n_bins)



# ====================================================================
# 3. FUNZIONE TEORICA DI MAXWELL-BOLTZMANN (3D)
# ====================================================================
def maxwell_boltzmann(v, T):
    norm = 4.0 * np.pi * (1.0 / (2.0 * np.pi * T))**(3.0/2.0)
    return norm * (v**2) * np.exp(-v**2 / (2.0 * T))

v_teorico = np.linspace(0, max(v), 500)

# ====================================================================
# 4. PLOT DELLA MEDIA PROGRESSIVA CON SBARRE D'ERRORE
# ====================================================================

fig, axes = plt.subplots(5, 4, figsize=(20, 16))
axes = axes.flatten() 


for i in range(n_blocks):
    ax = axes[i]
    
    # NORMALIZZAZIONE
    # Normalizziamo sia i valori dei bin che i loro rispettivi errori
    hist_normalized = hist_prog_blocks / (N_part * bin_size)
    err_normalized = err_prog_blocks / (N_part * bin_size)
    
    # PLOT DELL'ISTOGRAMMA CON SBARRE D'ERRORE
    # Aggiunto yerr per l'errore e capsize per i "tappini" orizzontali in cima alla barra
    ax.bar(v_blocks[i], hist_normalized[i], width=bin_size, align='center', 
           color='skyblue', edgecolor='black', alpha=0.7, 
           yerr=err_normalized[i], capsize=3, label='Dati MD Prog.')
    
    # PLOT DELLA CURVA TEORICA M-B
    t = temp_data[i]
    p_teorico = maxwell_boltzmann (v_teorico, t)
    ax.plot(v_teorico, p_teorico, color='red', linewidth=2.5, linestyle='--', 
            label=f'M-B ($T={t:.2f}$)')

    
    # DETTAGLI ESTETICI
    ax.set_title(f"Blocco {i + 1}", fontsize=12, fontweight='bold')
    ax.set_xlim(0, max(v) + bin_size)
    ax.set_ylim(0, max(hist_normalized[i]) + 0.05)
    ax.grid(True, alpha=0.3)
    ax.legend(loc='upper right', fontsize=8)
    
    if i >= 16: 
        ax.set_xlabel(r"Velocità $v$")
    if i % 4 == 0:  
        ax.set_ylabel(r"$p(v)$")

plt.suptitle("Evoluzione dell'Istogramma Progressivo delle Velocità (con errori)", 
             fontsize=20, y=0.98)

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()

```
