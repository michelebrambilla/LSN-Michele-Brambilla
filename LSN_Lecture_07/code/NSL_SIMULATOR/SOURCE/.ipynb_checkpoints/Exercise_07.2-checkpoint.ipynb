```python
block, energy = np.loadtxt("../OUTPUT/Exercise_07.2/potential_energyMC07.2.dat", usecols = (0, 1), unpack = True)

tmax = len(block)
chi_MC = []

max_lag = 250;

denominator = 1/tmax * np.sum(energy[:(tmax)]*energy[:(tmax)]) - (1/tmax*np.sum(energy[:(tmax)]))**2

for i in range (max_lag):
    upper_limit = tmax - i
    norm = 1/upper_limit

    first_term = norm*np.sum(energy[:upper_limit]*energy[i:(tmax)])
    second_term = norm*np.sum(energy[:upper_limit])*norm*np.sum(energy[i:(tmax)])

    chi_MC.append((first_term - second_term)/denominator)

plt.plot(block[:max_lag], chi_MC)
plt.grid(True)
#plt.yscale('log')
plt.title("Autocorrelation for MC")
plt.xlabel("Passo")
plt.ylabel("Autocorrelation")
plt.plot()
```

```python
block, energy = np.loadtxt("../OUTPUT/Exercise_07.2/potential_energyMD07.2.dat", usecols = (0, 1), unpack = True)

tmax = len(block)
chi = []

max_lag = 250;

denominator = 1/tmax * np.sum(energy[:(tmax)]*energy[:(tmax)]) - (1/tmax*np.sum(energy[:(tmax)]))**2

for i in range (max_lag):
    upper_limit = tmax - i
    norm = 1/upper_limit

    first_term = norm*np.sum(energy[:upper_limit]*energy[i:(tmax)])
    second_term = norm*np.sum(energy[:upper_limit])*norm*np.sum(energy[i:(tmax)])

    chi.append((first_term - second_term)/denominator)

plt.plot(block[:max_lag], chi)
plt.grid(True)
#plt.yscale('log')
plt.title("Autocorrelation for MD")
plt.xlabel("Passo")
plt.ylabel("Autocorrelation")
plt.plot()
```

Il tempo di autocorrelazione è più lungo per MD, e non decade esponenzialmente nei primissimi passi. Questo è dovuto al fatto che nel MC le particelle saltanodi qua e di là, mentre in MD seguono la dinamica molecolare, e hanno traiettorie continue. Il confronto tra le due curve di autocorrelazione è illustrato nel grafico successivo.

```python

fig, ax = plt.subplots(figsize=(10, 5), layout='constrained')
ax.plot(block[:max_lag], chi, label='MD autocorrelation') 
ax.plot(block[:max_lag], chi_MC, label='MC autocorrelation') 
ax.set_xlabel('Passo')  
ax.set_ylabel('Autocorrelazione') 
ax.grid(True)
ax.set_title("Confronto tra autocorrelazioni")  
ax.legend();

```

```python
block, energy = np.loadtxt("../OUTPUT/Exercise_07.2/potential_energyMC07.2.dat", usecols = (0, 1), unpack = True)

M = len(block)
L = np.arange(10, 5001, 10)
errors = []

for l in L:
    N = int(M/l)
    mean = []
    for i in range(N):
        mean.append(np.mean(energy[i*l:(i+1)*l]))
        
    variance = np.var(mean)
    errors.append(np.sqrt(variance/(N-1)))

plt.plot(L, errors)
plt.grid(True)
plt.title("Uncertainty on potential energy for MC")
plt.xlabel("Block length")
plt.ylabel("Uncertainty")
plt.plot()
```
