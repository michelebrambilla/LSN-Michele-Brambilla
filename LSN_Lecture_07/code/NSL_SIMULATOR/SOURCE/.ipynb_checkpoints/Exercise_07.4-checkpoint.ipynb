```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

block, energy = np.loadtxt("../OUTPUT/Exercise_07.4/potential_energyMD.dat", usecols = (0, 1), unpack = True)


plt.plot(block, energy)
plt.grid(True)
plt.title("Equilibration of potential energy for MD")
plt.xlabel("Step")
plt.ylabel("Potential energy")
plt.plot()

```

```python

block, t = np.loadtxt("../OUTPUT/Exercise_07.4/temperatureMD.dat", usecols = (0, 1), unpack = True)


plt.plot(block, t)
plt.grid(True)
plt.title("Equilibration of temperature for MD")
plt.xlabel("Step")
plt.ylabel("Temperature")
plt.plot()

```

```python
N = 100
L = 2000


block, energy, error = np.loadtxt("../OUTPUT/Exercise_07.4/total_energyMD07.4.dat", usecols = (0, 2, 3), unpack = True)


plt.errorbar(block, energy, yerr = error, fmt='o-', color='teal', 
             ecolor='darkorange', elinewidth=1.5, capsize=4, markersize=4, label='Stima Simulazione')
plt.grid(True)
plt.title("Evolution of total energy for MD")
plt.xlabel("Step")
plt.ylabel("Total energy")
plt.plot()
```

```python
N = 100
L = 2000


block, p, error = np.loadtxt("../OUTPUT/Exercise_07.4/pressureMD07.4.dat", usecols = (0, 2, 3), unpack = True)


plt.errorbar(block, p, yerr = error, fmt='o-', color='teal', 
             ecolor='darkorange', elinewidth=1.5, capsize=4, markersize=4, label='Stima Simulazione')
plt.grid(True)
plt.title("Evolution of pressure for MD")
plt.xlabel("Step")
plt.ylabel("Pressure")
plt.plot()
```

```python
d, g, err = np.loadtxt("../OUTPUT/Exercise_07.4/final_gofrMD07.4.dat", usecols = (0,1,2), unpack = True)


plt.errorbar(d, g, yerr = err, fmt='o-', color='teal', 
             ecolor='darkorange', elinewidth=1.5, capsize=4, markersize=4, label='Stima Simulazione')
plt.grid(True)
plt.title("g(r) for MD")
plt.xlabel("d (sigma)")
plt.ylabel("g(r)")
plt.plot()
```
